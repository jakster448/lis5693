## TASK 1: Visualize at least two different columns from the `seattle_df` dataframe

Here, we visualize the 'Checkouts' distribution using a bar chart and the 'UsageClass' counts also using a bar chart. These visualizations demonstrate basic data distribution and categorical counts.

```python
import altair as alt

# Visualization 1: Distribution of Checkouts
chart_checkouts_task1 = alt.Chart(seattle_df).mark_bar().encode(
    alt.X('Checkouts', bin=True),
    y='count()',
    tooltip=['Checkouts', 'count()']
).properties(
    title='Distribution of Checkouts'
).interactive()

display(chart_checkouts_task1)
```

```python
import altair as alt

# Visualization 2: Count of UsageClass
chart_usageclass_task1 = alt.Chart(seattle_df).mark_bar().encode(
    x=alt.X('UsageClass', sort='-y'),
    y='count()',
    tooltip=['UsageClass', 'count()']
).properties(
    title='Count of Materials by Usage Class'
).interactive()

display(chart_usageclass_task1)
```

## TASK 2: Use a different type of Altair mark for each visualization

For this task, we'll re-create visualizations using an area chart for 'Checkouts' distribution and a horizontal bar chart for 'MaterialType' counts, avoiding vertical bar graphs as requested.

```python
import altair as alt

# Visualization 1: Distribution of Checkouts using an Area Chart
chart_checkouts_area = alt.Chart(seattle_df).mark_area(opacity=0.7).encode(
    alt.X('Checkouts', bin=alt.Bin(maxbins=50)), # Use binning for distribution
    alt.Y('count()', stack=None),
    tooltip=['Checkouts', 'count()']
).properties(
    title='Distribution of Checkouts (Area Chart)'
).interactive()

display(chart_checkouts_area)
```

```python
import altair as alt

# Prepare data for MaterialType counts
material_type_counts = seattle_df['MaterialType'].value_counts().reset_index()
material_type_counts.columns = ['MaterialType', 'Count']

# Visualization 2: Count of MaterialType using a Horizontal Bar Chart
chart_material_type_bar = alt.Chart(material_type_counts).mark_bar().encode(
    alt.Y('MaterialType', sort='-x'), # Sort by count in descending order
    alt.X('Count'),
    tooltip=['MaterialType', 'Count']
).properties(
    title='Count of Materials by Type (Horizontal Bar Chart)'
).interactive()

display(chart_material_type_bar)
```

## TASK 3: Add interactivity to your graph (panning, zooming, tooltips, selections, binding, etc.)

Here, we create a stacked bar chart showing the number of checkouts by material type and year, adding advanced interactivity with a selection bound to the legend. This allows filtering by `CheckoutYear`.

```python
import altair as alt

# Group data for stacked bar chart
seattle_material_grouped = seattle_df.groupby(['MaterialType', 'CheckoutYear']).size().reset_index(name='counts')

# Create a selection for CheckoutYear, binding it to the legend
selection = alt.selection_point(fields=['CheckoutYear'], bind='legend')

interactive_chart_task3 = alt.Chart(seattle_material_grouped).mark_bar().encode(
    x=alt.X('MaterialType', sort='-y', title='Material Type'),
    y=alt.Y('counts', title='Number of Checkouts'),
    color=alt.Color('CheckoutYear:N', title='Checkout Year', sort='descending'),
    tooltip=['MaterialType', 'counts', 'CheckoutYear'],
    # Add interactivity: opacity condition based on selection
    opacity=alt.condition(selection, alt.value(1), alt.value(0.2)),
    order=alt.Order('CheckoutYear', sort='ascending') # Ensure proper stacking order
).add_params(
    selection
).properties(
    title='Number of Checkouts by Material Type and Year (Interactive)'
).interactive()

display(interactive_chart_task3)
```

## TASK 4: Horizontal concatenation of visualizations

We will combine the area chart for 'Checkouts' distribution and the horizontal bar chart for 'MaterialType' counts side-by-side using `alt.hconcat()`.

```python
import altair as alt

# Re-create Visualization 1: Distribution of Checkouts using an Area Chart
chart_checkouts_area_hconcat = alt.Chart(seattle_df).mark_area(opacity=0.7).encode(
    alt.X('Checkouts', bin=alt.Bin(maxbins=50)),
    alt.Y('count()', stack=None),
    tooltip=['Checkouts', 'count()']
).properties(
    title='Distribution of Checkouts (Area Chart)'
).interactive()

# Re-create Visualization 2: Count of MaterialType using a Horizontal Bar Chart
material_type_counts_hconcat = seattle_df['MaterialType'].value_counts().reset_index()
material_type_counts_hconcat.columns = ['MaterialType', 'Count']

chart_material_type_bar_hconcat = alt.Chart(material_type_counts_hconcat).mark_bar().encode(
    alt.Y('MaterialType', sort='-x'),
    alt.X('Count'),
    tooltip=['MaterialType', 'Count']
).properties(
    title='Count of Materials by Type (Horizontal Bar Chart)'
).interactive()

# Horizontally concatenate both charts
hconcatenated_chart_task4 = alt.hconcat(chart_checkouts_area_hconcat, chart_material_type_bar_hconcat).properties(
    title='Combined View: Checkouts Distribution and Material Type Counts'
)

display(hconcatenated_chart_task4)
```
