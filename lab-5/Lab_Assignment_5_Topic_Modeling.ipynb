Task 1: I chose "nutrition" as my dataset because it is a topic that connects health, economics, culture, and sustainability which makes it well-suited for topic modeling.

```python
import pandas as pd

url = "https://raw.githubusercontent.com/jakster448/lis5693/refs/heads/main/lab-5/lens-export.csv"
df = pd.read_csv(url)

df.head()
```

Task 2: I selected the abstract column for topic modeling because it contains the most detailed summary of each article’s content.

```python
text_data = df['Abstract']
text_data = text_data.dropna()
```

```python
import re
from nltk.corpus import stopwords

stop_words = set(stopwords.words('english'))

custom_stopwords = {
    'abstract', 'text', 'abstracttext',
    'introduction', 'method', 'methods',
    'results', 'conclusion', 'study',
    'using', 'based', 'analysis'
}

def preprocess(text):
    text = str(text).lower()

    # Fix merged words like "abstracttext"
    text = re.sub(r'(abstract)(text)', r'\1 \2', text)

    # Remove non-letters
    text = re.sub(r'[^a-zA-Z\s]', ' ', text)

    words = text.split()

    words = [
        w for w in words
        if w not in stop_words
        and w not in custom_stopwords
        and len(w) > 3
    ]

    return words
```

```python
!pip install gensim
from gensim import corpora
```

```python
processed_data = text_data.apply(preprocess)
dictionary = corpora.Dictionary(processed_data)
corpus = [dictionary.doc2bow(text) for text in processed_data]
```

```python
from gensim.models import LdaModel

lda_10 = LdaModel(corpus=corpus, id2word=dictionary, num_topics=10, passes=10)
lda_20 = LdaModel(corpus=corpus, id2word=dictionary, num_topics=20, passes=10)
```

Task 3: top 5 words of each topic

```python
print("LDA with 10 Topics:\n")
for i, topic in lda_10.print_topics(num_words=5):
    print(f"Topic {i}: {topic}")

print("\nLDA with 20 Topics:\n")
for i, topic in lda_20.print_topics(num_words=5):
    print(f"Topic {i}: {topic}")
```

Task 4: visualize topic results


*   Which topics appear most distinct?
The most distinct topics are those related to food security and agriculture, as they contain unique keywords and focus on different aspects of food systems.

*   Which topics overlap substantially?
Topics related to nutrition, diet, and health overlap significantly due to shared keywords such as “nutrition,” “health,” and “diet.”

*   Do the top keywords align with expectations?
Yes, the keywords align well with expectations, as they reflect major themes such as nutrition, health, and food systems.




```python
import matplotlib.pyplot as plt
from gensim.models import LdaModel

topic_nums = [5, 10, 15, 20]

for num in topic_nums:
    lda = LdaModel(corpus=corpus, id2word=dictionary, num_topics=num, passes=10)

    # Get topic distribution
    topic_counts = [0] * num
    for doc in corpus:
        topics = lda.get_document_topics(doc)
        for topic_id, prob in topics:
            topic_counts[topic_id] += prob

    # Plot
    plt.figure()
    plt.bar(range(num), topic_counts)
    plt.xlabel("Topics")
    plt.ylabel("Weight")
    plt.title(f"Topic Distribution ({num} Topics)")
    plt.show()
```
